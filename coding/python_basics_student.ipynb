## 1️⃣ 주석 (Comment) vs 코드 (Code)

파이썬 코드에는 두 가지 종류의 줄이 있어요.

- **주석(Comment)**: `#`으로 시작하는 줄. 파이썬이 **무시하고 실행하지 않아요.** 사람이 읽으라고 남기는 메모예요.

````python
# 이 줄은 주석입니다. 실행되지 않아요.
print("이 줄은 코드입니다. 실행됩니다!")  # 코드 뒤에 주석을 붙일 수도 있어요

# print("이 줄도 주석 처리되어 있어서 실행되지 않아요")
````

## 5️⃣ 함수 (Function) — 기본 개념

### 🚀 데모 3: 정답 확인 함수 만들기
함수로 "정답 확인기"를 만들어두면, 여러 문제에 재사용할 수 있어요. 이 함수는 잠시 후 최종 데모(퀴즈 게임)에서도 다시 사용됩니다!

````python
def check_answer(user_answer, correct_answer):
    # 대소문자와 앞뒤 공백을 무시하고 비교합니다
    return user_answer.strip().lower() == correct_answer.strip().lower()

print(check_answer("Apple", "apple"))     # True
print(check_answer(" APPLE ", "apple"))   # True
print(check_answer("banana", "apple"))    # False
````

## 6️⃣ 모듈(Module) · 라이브러리(Library) · 패키지(Package)

- **모듈(Module)**: 파이썬 코드가 담긴 파일 하나. `import 모듈이름`으로 가져와 씁니다.
- **라이브러리(Library)**: 여러 모듈을 모아놓은 것. (예: `random`은 파이썬에 기본 내장된 라이브러리예요)
- **패키지(Package)**: 여러 라이브러리를 체계적으로 묶어 배포하는 단위. 인터넷에서 `pip install 패키지이름`으로 설치해서 씁니다. (예: 영어 학습에 유용한 `gTTS`, `googletrans` 같은 패키지들!)

오늘은 파이썬에 **이미 내장되어 있어서 설치 없이 바로 쓸 수 있는** `random` 모듈을 사용해봅니다.

````python
import random

print(random.randint(1, 10))           # 1~10 사이 랜덤 정수
print(random.choice(["A", "B", "C"]))  # 리스트에서 랜덤으로 하나 선택
````

### 🎉 최종 데모: 랜덤 영어 단어 퀴즈 게임
오늘 배운 **모든 것**(주석, 입력/출력, 변수, 연산자, 함수, 모듈)을 합쳐서 진짜 퀴즈 앱을 만들어봤어요! 실행해서 몇 문제 풀어보세요.

> 💡 `for` 반복문과 리스트 인덱싱도 살짝 사용됐어요. 다음 시간에 배울 내용이니 지금은 몰라도 괜찮아요!

````python
#@markdown ### 🧠 랜덤 영어 단어 퀴즈 (Run! 버튼을 누르세요)
#@markdown 화면에 나오는 단어의 뜻을 한국어로 입력하면 채점해줍니다.

import random

# 단어와 뜻 (딕셔너리는 다음 시간에 배울 예정이라 오늘은 리스트 두 개로 간단히!)
word_list = ["apple", "book", "friend", "school", "happy"]
meaning_list = ["사과", "책", "친구", "학교", "행복한"]

def check_answer(user_answer, correct_answer):
    return user_answer.strip().lower() == correct_answer.strip().lower()

score = 0
total_questions = 3

for i in range(total_questions):
    index = random.randint(0, len(word_list) - 1)
    question_word = word_list[index]
    correct_meaning = meaning_list[index]

    user_answer = input(f"Q{i+1}. '{question_word}'의 뜻은 무엇일까요? (한국어로 입력) ")

    if check_answer(user_answer, correct_meaning):
        print("✅ 정답입니다!\n")
        score += 1
    else:
        print(f"❌ 아쉬워요! 정답은 '{correct_meaning}'였어요.\n")

percentage = score / total_questions * 100
print(f"🏁 최종 점수: {score}/{total_questions} ({percentage:.0f}점)")
````
